```python
import pandas as pd
```

```python
df_cp=pd.read_csv('codigos_postais.csv')
```

```python
df_cp.info()
```

```python
df_cp.head()
```

```python
# 1. Primeiro, temos de garantir que as colunas são tratadas como texto (string)
df_cp['num_cod_postal'] = df_cp['num_cod_postal'].astype(str)
df_cp['ext_cod_postal'] = df_cp['ext_cod_postal'].astype(str)
```

```python
# 2. Usar o zfill(3) na extensão para obrigar a ter sempre 3 dígitos
# Exemplo: '24' passa a '024', '5' passa a '005', '123' continua '123'
df_cp['ext_cod_postal'] = df_cp['ext_cod_postal'].str.zfill(3)
df_cp.head()
```

```python
#Criar uma nova coluna com os sete digitos juntos
df_cp['CP7'] = df_cp['num_cod_postal'] + '-' + df_cp['ext_cod_postal']
```

```python
df_cp.head()
```

```python
#Apenas deixo os valores que me interessam
df_cp= df_cp[['nome_localidade','num_cod_postal','ext_cod_postal','CP7']]
#Apenas os valores que começam por 4
df_cp = df_cp[df_cp['num_cod_postal'].str[0] == '4'].copy()
```

```python
df_cp.head()
```

```python
#Exportei o ficheiro para csv para dar merge
df_cp.to_csv('CodigoPostal.csv', index=False)
```
